## 2.1 Dasar Teori
**1. Pengenalan Struktur Perulangan for pada Python**

Struktur perulangan (looping) dalam Python adalah suatu struktur yang digunakan untuk mengulang eksekusi suatu blok kode atau untuk mengiterasi setiap elemen dari tipe data kolektif, seperti list. Python menyediakan dua jenis perulangan, yaitu perulangan for dan perulangan while. Perulangan for digunakan untuk menjalankan serangkaian pernyataan satu kali untuk setiap item dalam suatu urutan (sequence) atau jumlah pengulangan tertentu. Dengan memahami perulangan for, seorang programmer dapat menuliskan program yang lebih ringkas, tidak berulang-ulang, dan lebih mudah dipelihara.

**2. Konsep Struktur Perulangan for pada Python**

Perulangan for merupakan bagian dari program yang digunakan untuk mengulang suatu blok kode sebanyak elemen yang ada pada suatu objek yang dapat diiterasi (iterable). Berbeda dengan bahasa seperti C atau Pascal, yang perulangannya didasarkan pada deret angka atau pada langkah dan kondisi berhenti yang ditentukan sendiri oleh programmer, perulangan for pada Python bekerja dengan mengambil item satu per satu dari sebuah urutan sesuai urutan kemunculannya. Cara kerjanya lebih mirip metode iterator pada bahasa pemrograman berorientasi objek. Perulangan for ditulis dengan kata kunci for yang diikuti variabel penampung, kata kunci in, dan objek yang diiterasi, lalu diakhiri tanda titik dua (:). Variabel penampung akan menyimpan nilai elemen pada setiap putaran. Perulangan for juga tidak memerlukan variabel indeks yang disiapkan terlebih dahulu. Pernyataan yang menjadi isi perulangan wajib ditulis menjorok ke dalam (indentasi), karena indentasi inilah yang menandai bahwa pernyataan tersebut merupakan bagian dari blok perulangan. Jika indentasi tidak sesuai, program dapat menghasilkan error atau alur yang tidak sesuai harapan.

**3. Jenis Struktur Perulangan for dalam Python**

Ada 3 bentuk penerapan struktur perulangan for dalam Python, yaitu:
1. Perulangan for dengan data kolektif. Perulangan for dapat digunakan langsung pada tipe data kolektif, seperti list, tuple, string, dictionary, dan set. Pada list, tuple, dan set, perulangan mengambil setiap elemennya. Pada string, perulangan mengambil setiap karakter. Pada dictionary, perulangan secara bawaan mengambil key-nya, sedangkan value dapat diakses melalui key tersebut.
2. Perulangan for dengan fungsi range(). Fungsi range() digunakan untuk membuat deret angka yang berfungsi sebagai kontrol perulangan, sehingga blok kode dapat dijalankan sebanyak jumlah tertentu. Fungsi ini memiliki tiga bentuk penulisan, yaitu range(stop), range(start, stop), dan range(start, stop, step). Secara bawaan, deret dimulai dari angka 0 dan bertambah 1 pada setiap langkah. Nilai stop tidak pernah ikut menjadi bagian dari deret, sehingga range(5) menghasilkan angka 0 sampai 4. Nilai step dapat berupa bilangan negatif untuk menghasilkan deret menurun. Objek range berperilaku seperti list tetapi bukan list, karena tidak menyimpan seluruh nilainya di memori sekaligus. Jika diperlukan dalam bentuk list, objek tersebut dapat dikonversi dengan fungsi list().
3. Perulangan for bersarang (nested for). Perulangan for bersarang adalah perulangan yang berada di dalam perulangan lainnya. Perulangan di bagian dalam (inner loop) akan dijalankan penuh satu kali untuk setiap putaran perulangan di bagian luar (outer loop). Bentuk ini biasa digunakan untuk membuat pola, seperti deretan bintang, atau untuk mengolah data berdimensi lebih dari satu.

### 2.2.1 Program "jumlah-genap_2611532012"


```python
# Buat file dengan nama jumlah_genap_NIM.py
# Buat program untuk perulangan for dalam python
# Nama variabel ditambah 4 digit nim terakhir contoh : genap_1234
# Program ini menggunakan fungsi input()

ulang_2012 = int(input("Masukkan jumlah perulangan : "))

jumlah_2012 = 0
for i_2012 in range(1, ulang_2012 + 1):
    if i_2012 % 2 == 0:
        print(i_2012, end="")
        jumlah_2012 = jumlah_2012 + i_2012

        if  i_2012 < ulang_2012:
            print(" + ", end="")
        else:
            print(" = ", jumlah_2012, end="")
print()
print("Jumlah =", jumlah_2012)
```

```python
# Buat file dengan nama perulangan_for3_NIM.py
# Buat program untuk perulangan for dalam python
# Nama variabel ditambah 4 digit nim terakhir contoh : ulang_1234
# Program ini menggunakan fungsi input()

ulang_2012 = int(input("Masukkan jumlah perulangan : "))

jumlah_2012 = 0 
for i_2012 in range(1, ulang_2012 + 1):
    print(i_2012, end="")
    jumlah_2012 = jumlah_2012 + i_2012

    if i_2012 < ulang_2012:
        print(" + ", end="")
    else:
        print(" = ", end="")
print()
print("Jumlah =", jumlah_2012)
```
